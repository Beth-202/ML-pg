```python
# Prepare data1
data1 = {
    'id': ['0', '1', '2', '3', '4', '6', '8', '11', '12', '13'],
    'city': ['Tokyo', 'Osaka', 'Kyoto', 'Hokkaido', 'Tokyo', 'Tokyo', 'Osaka', 'Kyoto', 'Hokkaido', 'Tokyo'],
    'birth_year': [1990, 1989, 1992, 1997, 1982, 1991, 1988, 1990, 1995, 1981],
    'name': ['Hiroshi', 'Akiko', 'Yuki', 'Satoru', 'Steeve', 'Mituru', 'Aoi', 'Tarou', 'Suguru', 'Mitsuo'],
    'tuition': [2999,4898, 8888, 2047, 7868, 9429, 3389]
}
df1 = DataFrame(data1)
df1


```

```python
data2 = {
    'id': ['14', '15', '16', '17', '18', '19', '20'],
    'city': ['addis ababa', 'bahirdar', 'hawassa', 'mekelle', 'harar', 'asossa', 'gondar'],
    'birth_year': [1990, 2001,1992, 1920, 1993,1932, 1988],
    'name': ['Alemu', 'tariku', 'kidist', 'helen', 'kaleabb', 'samrawit', 'Selam'],
    'grade':['A', 'A', 'B', 'B', 'C' , 'A', 'B' ],
    'major': ['biology', 'chemistry', 'biology', 'engineering', 'medicine', 'math', 'marketing']
    }
df2 = DataFrame(data2)
df2
```

```python
#inner join
pd.merge(df1, df2, on='id')
```

```python
pd.merge(df1, df2, left_on="id", right_on="city")
```
